```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

class LogisticRegressionOneVsAll:
    def __init__(self, num_classes=10, learning_rate=0.1, num_iterations=500, lambda_reg=0.1):
        self.num_classes = num_classes
        self.learning_rate = learning_rate
        self.num_iterations = num_iterations
        self.lambda_reg = lambda_reg
        self.thetas = None

    def sigmoid(self, z):
        z = np.clip(z, -500, 500)
        return 1 / (1 + np.exp(-z))

    def compute_cost(self, X, y, theta):
        m = len(y)
        h = self.sigmoid(X @ theta)
        cost = (-1/m) * (y.T @ np.log(h + 1e-15) + (1-y).T @ np.log(1-h + 1e-15))
        reg = (self.lambda_reg/(2*m)) * np.sum(theta[1:]**2)
        return cost + reg


def train_binary_classifier(self, X, y, digit):
    # Начальные веса
    theta = np.zeros(X.shape[1])  # [0, 0, ..., 0]

    for i in range(self.num_iterations):  # 500 раз

        # ВЫЧИСЛЯЕМ ГРАДИЕНТ
        m = len(y)
        h = self.sigmoid(X @ theta)           # предсказания
        gradient = (1/m) * (X.T @ (h - y))    # основной градиент
        reg_gradient = (self.lambda_reg/m) * theta  # регуляризация
        reg_gradient[0] = 0
        gradient += reg_gradient

        # ОБНОВЛЯЕМ ВЕСА (ОДНА СТРОКА!)
        theta = theta - self.learning_rate * gradient

    return theta

    def fit(self, X, y):
        X_with_bias = np.c_[np.ones(X.shape[0]), X]
        self.thetas = np.zeros((self.num_classes, X_with_bias.shape[1]))

        print("\nНачало обучения 10 классификаторов one-vs-all:")
        for digit in range(self.num_classes):
            print(f"\nОбучение классификатора для цифры {digit}:")
            self.thetas[digit] = self.train_binary_classifier(X_with_bias, y, digit)
        print("\nОбучение завершено!")

    def predict_proba(self, X):
        X_with_bias = np.c_[np.ones(X.shape[0]), X]
        probabilities = self.sigmoid(X_with_bias @ self.thetas.T)
        return probabilities

    def predict(self, X):
        probabilities = self.predict_proba(X)
        return np.argmax(probabilities, axis=1)

def normalize_features(X):
    """Нормировка признаков (стандартизация)"""
    mean = np.mean(X, axis=0)
    std = np.std(X, axis=0)
    std[std == 0] = 1
    X_normalized = (X - mean) / std
    return X_normalized, mean, std

def load_and_prepare_data():
    """Загрузка данных с правильной обработкой меток"""
    print("Загрузка данных...")

    # Загрузка признаков
    X = pd.read_csv('data3_X.txt', header=None, sep='\s+').values

    # Загрузка меток
    y = pd.read_csv('data3_Y.txt', header=None).values.flatten()

    # ВАЖНО: Заменяем 10.0 на 0 (в вашем файле 10 означает 0)
    # Также убеждаемся, что все метки в диапазоне 0-9
    y = y.astype(int)
    y[y == 10] = 0  # В ваших данных 10 означает цифру 0

    print(f"Размер матрицы признаков X: {X.shape}")
    print(f"Размер вектора ответов y: {y.shape}")
    print(f"Уникальные метки после исправления: {np.unique(y)}")
    print(f"Распределение меток:")
    for digit in range(10):
        count = np.sum(y == digit)
        print(f"  Цифра {digit}: {count} примеров ({count/len(y)*100:.1f}%)")

    return X, y

def shuffle_data(X, y):
    """Перемешивание данных"""
    np.random.seed(42)
    indices = np.random.permutation(len(y))
    return X[indices], y[indices]

def split_train_test(X, y, train_ratio=0.8):
    """Разделение на обучающую и тестовую выборки"""
    split_idx = int(len(y) * train_ratio)
    X_train, X_test = X[:split_idx], X[split_idx:]
    y_train, y_test = y[:split_idx], y[split_idx:]

    print(f"\nРазделение данных:")
    print(f"Обучающая выборка: {X_train.shape[0]} примеров")
    print(f"Тестовая выборка: {X_test.shape[0]} примеров")

    return X_train, X_test, y_train, y_test

def display_digit(X, index, title="Изображение цифры"):
    """Отображение цифры из вектора признаков"""
    digit_image = X[index].reshape(20, 20)
    plt.figure(figsize=(4, 4))
    plt.imshow(digit_image.T, cmap='gray', interpolation='bilinear')
    plt.title(title, fontsize=14, fontweight='bold')
    plt.axis('off')
    plt.show()

def calculate_accuracy(y_true, y_pred):
    """Вычисление точности"""
    return np.mean(y_true == y_pred) * 100

def confusion_matrix(y_true, y_pred, num_classes=10):
    """Построение матрицы ошибок"""
    cm = np.zeros((num_classes, num_classes), dtype=int)
    for true, pred in zip(y_true, y_pred):
        cm[true, pred] += 1
    return cm

def plot_confusion_matrix(cm):
    """Отображение матрицы ошибок"""
    plt.figure(figsize=(10, 8))
    plt.imshow(cm, interpolation='nearest', cmap='Blues')
    plt.title('Матрица ошибок', fontsize=16, fontweight='bold')
    plt.colorbar()
    plt.xlabel('Предсказанные метки', fontsize=12)
    plt.ylabel('Истинные метки', fontsize=12)

    # Добавляем значения в ячейки
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(j, i, str(cm[i, j]), ha='center', va='center')

    plt.xticks(range(10))
    plt.yticks(range(10))
    plt.show()

def plot_training_results(y_test, y_pred, X_test, y_test_original=None):
    """Визуализация результатов"""
    accuracy = calculate_accuracy(y_test, y_pred)

    # Показываем примеры для каждой цифры
    fig, axes = plt.subplots(2, 5, figsize=(15, 7))
    axes = axes.flatten()

    for digit in range(10):
        # Находим правильно распознанный пример для каждой цифры
        mask = (y_test == digit) & (y_pred == digit)
        if np.any(mask):
            idx = np.where(mask)[0][0]
        else:
            # Если нет правильно распознанных, берем любой
            mask2 = (y_test == digit)
            if np.any(mask2):
                idx = np.where(mask2)[0][0]
            else:
                continue

        img = X_test[idx].reshape(20, 20)
        axes[digit].imshow(img.T, cmap='gray', interpolation='bilinear')
        axes[digit].set_title(f'Цифра: {digit}', fontsize=12, fontweight='bold')
        axes[digit].axis('off')

    plt.suptitle(f'Примеры изображений из тестовой выборки\n(Точность: {accuracy:.1f}%)',
                fontsize=16, fontweight='bold')
    plt.tight_layout()
    plt.show()

def main():
    print("="*60)
    print("Распознавание рукописных цифр")
    print("Логистическая регрессия (One-vs-All)")
    print("="*60)

    # 1. Загрузка данных
    X, y = load_and_prepare_data()

    # 2. Перемешивание данных
    X, y = shuffle_data(X, y)

    # 3. Разделение на обучающую и тестовую выборки
    X_train, X_test, y_train, y_test = split_train_test(X, y, train_ratio=0.8)

    # 4. Нормировка признаков (данные уже нормализованы, но для единообразия)
    print("\nНормировка признаков...")
    X_train_norm, mean, std = normalize_features(X_train)
    X_test_norm = (X_test - mean) / std

    print(f"После нормировки: среднее ≈ {np.mean(X_train_norm):.6f}, std ≈ {np.std(X_train_norm):.6f}")

    # 5. Обучение модели (оптимизированные параметры для ваших данных)
    model = LogisticRegressionOneVsAll(
        num_classes=10,
        learning_rate=0.5,      # Увеличен для более быстрой сходимости
        num_iterations=500,      # Больше итераций для лучшего обучения
        lambda_reg=0.01          # Меньше регуляризации
    )

    model.fit(X_train_norm, y_train)

    # 6. Предсказания на тестовой выборке
    print("\nПредсказание на тестовой выборке...")
    y_pred = model.predict(X_test_norm)

    # 7. Оценка качества
    accuracy = calculate_accuracy(y_test, y_pred)
    print(f"\n{'='*60}")
    print(f"ТОЧНОСТЬ НА ТЕСТОВОЙ ВЫБОРКЕ: {accuracy:.2f}%")
    print(f"{'='*60}")

    # 8. Детальная статистика по каждой цифре
    print("\nДетальная статистика по каждой цифре:")
    print("-" * 50)
    for digit in range(10):
        mask = (y_test == digit)
        if np.sum(mask) > 0:
            acc_digit = calculate_accuracy(y_test[mask], y_pred[mask])
            correct = np.sum(y_test[mask] == y_pred[mask])
            total = np.sum(mask)
            print(f"  Цифра {digit}: {acc_digit:.1f}% ({correct}/{total} примеров)")

    # 9. Матрица ошибок
    cm = confusion_matrix(y_test, y_pred)
    print("\nМатрица ошибок:")
    print(cm)
    plot_confusion_matrix(cm)

    # 10. Визуализация примеров
    plot_training_results(y_test, y_pred, X_test)

    # 11. Демонстрация предсказаний для случайных примеров
    print("\n" + "="*60)
    print("Демонстрация предсказаний для случайных примеров:")
    print("="*60)

    num_examples = 5
    random_indices = np.random.choice(len(X_test), min(num_examples, len(X_test)), replace=False)

    for idx in random_indices:
        true_digit = int(y_test[idx])
        pred_digit = int(y_pred[idx])
        proba = model.predict_proba(X_test_norm[idx:idx+1])[0]

        # Определяем цвет заголовка
        color = 'green' if true_digit == pred_digit else 'red'

        display_digit(X_test, idx,
                     title=f"Истинная: {true_digit} → Предсказанная: {pred_digit}")

        print(f"Вероятности для каждого класса:")
        for i in range(10):
            bar = "█" * int(proba[i] * 50)
            print(f"  Цифра {i}: {proba[i]:.3f} {bar}")

        # Показываем максимум вероятности
        max_prob = np.max(proba)
        print(f"Уверенность: {max_prob:.1%}")
        print("-" * 50)

    # 12. Анализ ошибок
    errors = np.where(y_pred != y_test)[0]
    if len(errors) > 0:
        print(f"\nВсего ошибок: {len(errors)} из {len(y_test)} ({len(errors)/len(y_test)*100:.1f}%)")

        print("\nПримеры ошибок классификации:")
        num_errors_to_show = min(3, len(errors))
        for error_idx in errors[:num_errors_to_show]:
            true_digit = int(y_test[error_idx])
            pred_digit = int(y_pred[error_idx])
            display_digit(X_test, error_idx,
                         title=f"ОШИБКА! Истинная: {true_digit}, Предсказанная: {pred_digit}")

            proba = model.predict_proba(X_test_norm[error_idx:error_idx+1])[0]
            print(f"Вероятности для классов:")
            for i in range(10):
                if i == true_digit:
                    print(f"  Цифра {i}: {proba[i]:.3f} ★ (истинная)")
                elif i == pred_digit:
                    print(f"  Цифра {i}: {proba[i]:.3f} ★ (предсказанная)")
                else:
                    print(f"  Цифра {i}: {proba[i]:.3f}")
            print("-" * 50)

if __name__ == "__main__":
    main()
```
